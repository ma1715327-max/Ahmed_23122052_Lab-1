```python
print("===== WORLD TRAVEL ADVISOR =====")

country = input("Which country do you want to visit? ")
money = int(input("Enter your travel budget in PKR: "))

if country.lower() == "turkey":
    if money >= 150000:
        print("Great! Turkey is suitable for your budget.")
        print("Places to visit: Istanbul, Cappadocia and Antalya.")
    else:
        print("Your budget may not be enough for Turkey.")

if country.lower() == "malaysia":
    if money >= 120000:
        print("You can plan a trip to Malaysia.")
        print("Places to visit: Kuala Lumpur, Langkawi and Penang.")
    else:
        print("You need a higher budget for Malaysia.")

if country.lower() == "dubai":
    if money >= 180000:
        print("Dubai can be a good choice for your trip.")
        print("Places to visit: Burj Khalifa, Dubai Mall and Jumeirah.")
    else:
        print("Your budget is low for a Dubai trip.")

else:
    print("This country is not in our travel list.")
    print("Available options: Turkey, Malaysia or Dubai.")

print("Thanks for using World Travel Advisor!")
```
